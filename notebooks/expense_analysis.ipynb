# Expense Data Exploration
This notebook demonstrates Pandas, Matplotlib and Seaborn on the synthetic CSV.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../data/sample_expenses.csv", parse_dates=["date"])
df.head()
```

```python
df.groupby("category")["amount"].sum().sort_values(ascending=False)
```

```python
monthly = df.assign(month=df.date.dt.to_period("M").astype(str)).groupby("month").amount.sum()
monthly.plot(kind="line", marker="o", title="Monthly spending trend")
plt.ylabel("Amount (INR)"); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df.groupby("category",as_index=False).amount.sum(), x="category", y="amount")
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
